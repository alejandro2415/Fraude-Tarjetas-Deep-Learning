```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight

# 1. Cargar el archivo extraído en un DataFrame de Pandas
df = pd.read_csv('creditcard.csv')

# 2. Escalar variables numéricas 'Time' y 'Amount' para que la red neuronal trabaje mejor
scaler = StandardScaler()
df['Amount'] = scaler.fit_transform(df['Amount'].values.reshape(-1, 1))
df['Time'] = scaler.fit_transform(df['Time'].values.reshape(-1, 1))

# 3. Separar las variables de entrada (X) y la etiqueta objetivo (y)
X = df.drop(columns=['Class'])
y = df['Class']

# 4. Dividir en conjuntos de entrenamiento (80%) y prueba (20%) manteniendo la proporción de fraudes
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# 5. Calcular los pesos de clase para penalizar con más fuerza si el modelo ignora un fraude
pesos = compute_class_weight(class_weight='balanced', classes=np.unique(y_train), y=y_train)
class_weights = {0: pesos[0], 1: pesos[1]}

# Mostrar resumen en pantalla
print("--- PREPROCESAMIENTO COMPLETADO ---")
print(f"Registros para entrenar la red: {X_train.shape[0]}")
print(f"Registros para validar/evaluar: {X_test.shape[0]}")
print(f"Peso asignado a transacciones Normales (Clase 0): {class_weights[0]:.4f}")
print(f"Peso asignado a transacciones con Fraude (Clase 1): {class_weights[1]:.4f}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# 1. Definición de la arquitectura secuencial de la Red Neuronal
model = Sequential([
    # Capa de entrada con 30 nodos (uno por cada variable) y función ReLU
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    # Capa Dropout para apagar aleatoriamente el 30% de las neuronas y evitar sobreajuste
    Dropout(0.3),
    # Capa oculta con 16 nodos
    Dense(16, activation='relu'),
    Dropout(0.2),
    # Capa de salida con 1 nodo y activación Sigmoid (devuelve un valor entre 0 y 1)
    Dense(1, activation='sigmoid')
])

# 2. Compilar el modelo definiendo la pérdida y las métricas a monitorear
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy', tf.keras.metrics.Recall(), tf.keras.metrics.Precision()])

# 3. Iniciar el entrenamiento pasando los datos y los pesos calculados antes
print("Iniciando el entrenamiento de la Red Neuronal...")
history = model.fit(X_train, y_train,
                    epochs=15,
                    batch_size=2048,
                    validation_split=0.1,
                    class_weight=class_weights,
                    verbose=1)
print("\n--- ¡ENTRENAMIENTO COMPLETADO CON ÉXITO! ---")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# 1. Realizar las predicciones con el conjunto de prueba (X_test)
y_pred_prob = model.predict(X_test)

# 2. Umbral de decisión: si la probabilidad es mayor al 50%, se clasifica como fraude (1)
y_pred = (y_pred_prob > 0.5).astype(int)

# 3. Imprimir el reporte formal de clasificación (Precision, Recall, F1-Score)
print("\n--- REPORTE DE CLASIFICACIÓN ---")
print(classification_report(y_test, y_pred, target_names=['Normal', 'Fraude']))

# 4. Calcular y graficar la Matriz de Confusión
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicho Normal', 'Predicho Fraude'],
            yticklabels=['Real Normal', 'Real Fraude'])

plt.title('Matriz de Confusión - Detección de Fraude con Red Neuronal')
plt.ylabel('Clase Real')
plt.xlabel('Clase Predicha')
plt.tight_layout()
plt.show()
```
